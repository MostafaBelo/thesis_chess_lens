```python
with open("/mnt/D/University/Thesis_Dataset/Dataset/FENs_old.txt", "r") as f:
    data = f.read()
```

```python
data = data.strip().split("\n")
data
```

```python
data2 = []
for line in data:
    index = line.find(".#")
    id = line[:index]
    fen = line[index+2:]
    if int(id) >= 120 and int(id) <= 178:
        ext = "png"
    else:
        ext = "jpg"
    orientation = "l" if int(id) in list(range(101, 273)) else "r"

    fen = fen.strip("(").strip(")")

    image_path = f"Images/image_{id}.{ext}"
    new_line = f"{id}.#({image_path})({fen})({orientation})"
    data2.append(new_line)
data2
```

```python
with open("/mnt/D/University/Thesis_Dataset/Dataset/labels.txt", "w") as f:
    f.write("\n".join(data2))
```

```python
import re
import json
import os
```

```python
with open("/mnt/D/University/Thesis_Dataset/Dataset/labels.txt", "r") as f:
    data3 = f.read()
    data3 = data3.strip().split("\n")
    # f.write("\n".join(data2))
```

```python
with open("/mnt/D/University/Thesis/src/Dataset/DatasetCornerAnnotator/chessboard_corners_all.json", "r") as f:
    corners_data = json.loads(f.read())

new_corners_data = {}
for file_name in list(corners_data.keys()):
    img_id = os.path.basename(file_name).strip(".jpg").strip(".png").strip("image_")
    res = []
    res += corners_data[file_name]["top-left"]
    res += corners_data[file_name]["top-right"]
    res += corners_data[file_name]["bottom-right"]
    res += corners_data[file_name]["bottom-left"]
    new_corners_data[img_id] = ",".join(map(lambda x: str(x), res))
```

```python
data4 = []
for line in data3:
    index = line.find(".#")
    id = line[:index]

    data4.append(line + f"({new_corners_data[id]})")
```

```python
with open("/mnt/D/University/Thesis_Dataset/Dataset/labels2.txt", "w") as f:
    f.write("\n".join(data4))
```
