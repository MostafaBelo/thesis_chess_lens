```python
import json
import cv2
from matplotlib import pyplot as plt

import os
```

```python
with open("./chessboard_corners_all.json", "r") as f:
    all_corners = json.loads(f.read())
```

```python
def plot_circle(point, img):
    img_copy = img.copy()
    cv2.circle(img_copy, point, 5, (0, 0, 255), -1)
    return img_copy
```

```python
files = list(all_corners.keys())
```

```python
i = 300
```

```python
img = cv2.imread(files[i])
for corner_name in all_corners[files[i]].keys():
    img = plot_circle(all_corners[files[i]][corner_name], img)
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.imshow(img);

print(i, files[i])
i += 1
```
